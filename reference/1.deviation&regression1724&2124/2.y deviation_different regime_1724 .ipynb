```python
import pandas as pd

# 读取 Excel 文件
file_path = 'rolling_regression_results_regime2.xlsx'  # 替换为你的 Excel 文件路径
df = pd.read_excel(file_path)

# 计算 y estimate 和 deviation
df['y_estimate'] = (
    df['Coef_X1'].shift(1) * (df['CL/CP']) +
    df['Coef_X2'].shift(1) * (df['NTt+1/CP'])+
    df['Coef_Constant'].shift(1)
)

df['deviation'] = df['CPt+1/CP'] - df['y_estimate']

# 将结果写入新的 Excel 文件
output_file_path = 'rolling_regression_deviation_regime2.xlsx'  # 替换为输出文件路径
df.to_excel(output_file_path, index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取数据
df2 = pd.read_excel('rolling_regression_deviation_regime2.xlsx')
df3 = pd.read_excel('rolling_regression_deviation_regime3.xlsx')

# 将Date列转换为datetime格式
df2['Date'] = pd.to_datetime(df2['Date'])
df3['Date'] = pd.to_datetime(df3['Date'])

# 创建一个合并的数据框
combined_df = pd.DataFrame()

# 选择数据并合并
#因为full time能算的rolling window就算，最早也要从以subperiod第31天才开始有rolling window的结果

segment1 = df2[(df2['Date'] >= '2017-02-20') & (df2['Date'] <= '2017-05-25')]
if len(segment1) > 30:
    segment1.iloc[:30, segment1.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment1])

segment2 = df3[(df3['Date'] >= '2017-05-26') & (df3['Date'] <= '2018-01-08')]
if len(segment2) > 30:
    segment2.iloc[:30, segment2.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment2])


segment3 = df2[(df2['Date'] >= '2018-01-09') & (df2['Date'] <= '2018-08-23')]
if len(segment3) > 30:
    segment3.iloc[:30, segment3.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment3])


segment4 = df3[(df3['Date'] >= '2018-08-24') & (df3['Date'] <= '2020-10-26')]
if len(segment4) > 30:
    segment4.iloc[:30, segment4.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment4])


segment5 = df2[(df2['Date'] >= '2020-10-27') & (df2['Date'] <= '2021-12-08')]
if len(segment5) > 30:
    segment5.iloc[:30, segment5.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment5])

# 2021-12-09到2022-03-17用df3的数据
segment2 = df3[(df3['Date'] >= '2021-12-09') & (df3['Date'] <= '2022-03-27')]
if len(segment2) > 30:
    segment2.iloc[:30, segment2.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment2])

# 2022-03-18到2022-08-41用df2的数据
segment3 = df2[(df2['Date'] >= '2022-03-18') & (df2['Date'] <= '2022-08-31')]
if len(segment3) > 30:
    segment3.iloc[:30, segment3.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment3])

# 2022-09-01到2022-12-04用df3的数据
segment4 = df3[(df3['Date'] >= '2022-09-01') & (df3['Date'] <= '2022-12-04')]
if len(segment4) > 30:
    segment4.iloc[:30, segment4.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment4])

# 2022-12-05到2023-6-28用df2的数据
segment5 = df2[(df2['Date'] >= '2022-12-05') & (df2['Date'] <= '2023-06-28')]
if len(segment5) > 30:
    segment5.iloc[:30, segment5.columns.get_loc('deviation')] = None
combined_df = pd.concat([combined_df, segment5])



# 绘制折线图
plt.figure(figsize=(12, 6))

plt.plot(combined_df['Date'], combined_df['deviation'], label='Combined Deviation', color='blue')
plt.legend()

# 添加垂直线
highlight_date = '2017-5-25'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2018-01-08'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2018-08-23'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2020-10-26'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2021-12-08'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2022-03-17'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2022-08-31'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2022-12-04'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)

highlight_date = '2023-6-28'
plt.axvline(pd.to_datetime(highlight_date), color='green', linestyle='--', label='Highlight Date')
plt.text(pd.to_datetime(highlight_date), plt.ylim()[1] * 0.1, highlight_date,
         horizontalalignment='center', color='black', rotation=90)


plt.title('Combined Deviation over Time')
plt.xlabel('Dates')
plt.ylabel('Deviation')
plt.xticks(rotation=45)

plt.grid()

# 保存图形
plt.savefig('17-24combined_deviation_plot.png', bbox_inches='tight')

# 显示图形
plt.tight_layout()
plt.show()
```
